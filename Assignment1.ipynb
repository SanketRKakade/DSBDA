```python
import numpy as np

def standard_scalar(data):
    mean = np.mean(data)
    std_dev = np.std(data)
    scaled_data = [(x-mean)/std_dev for x in data]
    return scaled_data

std_data = standard_scalar(df['Fare'])
np.mean(std_data)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
normalised_data = scaler.fit_transform(df['Fare'].values.reshape(-1,1))
normalised_data
normalised_data.mean()
```

```python
import numpy as np
def minmaxscaler (data):
    min = np.min(data)
    max = np.max(data)
    nor_data = [(x-min)/(max-min) for x in data]
    return nor_data

scaler = minmaxscaler(df['Fare'])
print(np.mean(scaler))
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Assuming your DataFrame is loaded as 'df'

# Apply Label Encoding to the 'Sex' column
label_encoder = LabelEncoder()
df['Sex_LabelEncoded'] = label_encoder.fit_transform(df['Sex'])

print("DataFrame with 'Sex' Label Encoded:")
print(df[['Sex', 'Sex_LabelEncoded']].head())
```

```python
# do only when asked

import pandas as pd

# Assuming your DataFrame is loaded as 'df'

# Apply Dummy Coding to 'Sex' and 'Embarked'
# drop_first=True removes the first category column to avoid multicollinearity
df_dummies = pd.get_dummies(df[['Sex', 'Embarked']], drop_first=True)

print("DataFrame with Dummy Coded 'Sex' and 'Embarked':")
print(df_dummies.head())

# You can concatenate these new columns with your original DataFrame
df = pd.concat([df, df_dummies], axis=1)
print("\nOriginal DataFrame with Dummy Coded columns:")
print(df.head())
```

```python
#do if asked 

import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Assuming your DataFrame is loaded as 'df'

# Handle missing values in 'Embarked' before One-Hot Encoding
df['Embarked'] = df['Embarked'].fillna('Unknown') # Replace NaN with a placeholder

# Apply One-Hot Encoding to 'Sex' and 'Embarked'
onehot_encoder = OneHotEncoder(sparse_output=False) # sparse_output=False returns a numpy array

# Fit and transform the selected columns
# Use .values.reshape(-1, 1) if encoding a single column
onehot_encoded_data = onehot_encoder.fit_transform(df[['Sex', 'Embarked']])

# Create a DataFrame from the one-hot encoded data
# Get the new column names
feature_names = onehot_encoder.get_feature_names_out(['Sex', 'Embarked'])
df_onehot = pd.DataFrame(onehot_encoded_data, columns=feature_names, index=df.index)

print("DataFrame with One-Hot Encoded 'Sex' and 'Embarked':")
print(df_onehot.head())

# You can concatenate these new columns with your original DataFrame
# df = pd.concat([df, df_onehot], axis=1)
# print("\nOriginal DataFrame with One-Hot Encoded columns:")
# print(df.head())
```
