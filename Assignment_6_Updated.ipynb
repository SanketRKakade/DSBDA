## 📊 Histograms of All Features

```python

data.hist(figsize=(10, 8), color='skyblue', edgecolor='black')
plt.suptitle('Feature Distributions - Histograms')
plt.show()
```

## 📦 Boxplots of All Features

```python

plt.figure(figsize=(12, 6))
for i, column in enumerate(['sl', 'sw', 'pl', 'pw']):
    plt.subplot(1, 4, i + 1)
    sns.boxplot(y=data[column], color='lightgreen')
    plt.title(f'Boxplot of {column}')
plt.tight_layout()
plt.show()
```

## 📦 Boxplots by Species

```python

plt.figure(figsize=(12, 6))
sns.boxplot(data=data, x='variety', y='pl')
plt.title('Petal Length Distribution by Species')
plt.show()
```

## 🔗 Pairplot for Relationships

```python

sns.pairplot(data, hue='variety')
plt.suptitle("Pairwise Feature Relationships by Species", y=1.02)
plt.show()
```

## 🔥 Correlation Heatmap

```python

plt.figure(figsize=(8, 6))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='Blues')
plt.title("Feature Correlation Heatmap")
plt.show()
```
