```python
from nltk.stem import PorterStemmer , WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()
```

```python
def stem_lem (tokens):
    lem_words = [lemmatizer.lemmatize(word) for word in tokens]
    stem_words = [stemmer.stem(word) for word in tokens]
    return lem_words,stem_words
```

```python
lem_words , stem_words = stem_lem(token_words)
```

```python
def caltf(doc):
    tokens=nltk.word_tokenize(doc)
    temp=dict()
    for word in tokens:
        temp[word]=tokens.count(word)
    tf=dict()
    for word,cnt in temp.items():
        tf[word]=cnt/(len(tokens))
    return tf
```

```python
tf1=caltf(doc1)
tf2=caltf(doc2)
```

```python
print(tf1)
```

```python
print(tf2)
```

```python
tokens1=nltk.word_tokenize(doc1)
tokens2=nltk.word_tokenize(doc2)
def cal_idf():
    all_tokens=set(tokens1+tokens2)
    idf=dict()
    for word in all_tokens:
        f=0
        if word in tokens1:
            f+=1
        if word in tokens2:
            f+=1
        if f>0:
            idf[word]=math.log(2/f)
        else:
            idf[word]=0
    return idf
```

```python
idf=cal_idf()
print(idf)
```

```python
tf_idf_1=dict()
tf_idf_2=dict()
for word in tokens1:
    tf_idf_1[word]=(tf1[word]*idf[word])
for word in tokens2:
    tf_idf_2[word]=(tf2[word]*idf[word])
```
